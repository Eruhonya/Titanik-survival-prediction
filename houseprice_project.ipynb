```python
# Function

# Calculate house percentage and mean price per category (via Series)
def calculate_mean_traget_per_category(df, var):
    total_houses = len(df)
    temp_df = pd.Series(df[var].value_counts() / total_houses).reset_index()
    temp_df.columns = [var, 'perc_houses']
    
    temp_df = temp_df.merge(
        df.groupby([var])['SalePrice'].mean().reset_index(),
        on=var, how='left'
    )
    return temp_df

# Calculate category frequency and mean price (via DataFrame)
def calculate_mean_target_per_category(df, var):
    total_houses = len(df)
    temp_df = pd.DataFrame(df[var].value_counts()).reset_index()
    temp_df.columns = [var, 'count_houses']
    temp_df['perc_houses'] = (temp_df['count_houses'] / total_houses)
    
    temp_df = temp_df.merge(
        df.groupby([var])['SalePrice'].mean().reset_index(),
        on=var, how='left'
    )
    return temp_df

# Visualize house percentage (bar) and mean price (line) per category
def plot_categories(df, var):
    fig, ax = plt.subplots(figsize=(6,3))
    plt.xticks(df.index, df[var], rotation=90)

    ax2 = ax.twinx()
    ax.bar(df.index, df['perc_houses'], color='lightgrey')
    ax2.plot(df.index, df['SalePrice'], color='green', label='Second')
    ax.axhline(y=0.05, color='red')  # 5% rare category threshold
    ax.set_ylabel('Percentage of houses per category')
    ax.set_xlabel(var)
    ax2.set_ylabel('Average Sale price per category')
    plt.show()

# Comprehensive categorical feature analysis (stats + boxplot + charts)
def feature_to_look(col):
    temp_df = calculate_mean_target_per_category(data, col)
    print(temp_df)
    print(f'\n{(data[col].isna().mean() * 100).round(2)}% - missing value.\n')
    sns.boxplot(x=col, y='SalePrice', data=data)
    plot_categories(temp_df, col)

# Statistical and graphical analysis for continuous variables (4 subplots)
def plot_variable_analysis(df, variables, target):
    for var in variables:
        df_var = df[var].dropna()
        target_filtered = df[target].loc[df_var.index]
        
        missing_pct = (df[var].isna().sum() / len(df)) * 100
        mean_val, median_val = df_var.mean(), df_var.median()
        mode_vals = df_var.mode()
        mode_val = mode_vals.iloc[0] if not mode_vals.empty else np.nan
        std_val, var_val = df_var.std(), df_var.var()
        
        stats_text = (
            f"Mean: {mean_val:.2f} | Median: {median_val:.2f} | Mode: {mode_val:.2f}\n"
            f"Std: {std_val:.2f} | Variance: {var_val:.2f}| Missing: {missing_pct:.2f}\n"
        )
        
        plt.figure(figsize=(10, 6))
        
        # Subplot 1: Distribution (Histogram + KDE)
        plt.subplot(2, 2, 1)
        sns.histplot(df_var, kde=True)
        plt.axvline(mean_val, color='r', linestyle='--', linewidth=1, label='Mean')
        plt.axvline(median_val, color='g', linestyle='-', linewidth=1, label='Median')
        plt.axvline(mode_val, color='b', linestyle=':', linewidth=1, label='Mode')
        plt.legend()
        plt.title(f'Distribution of {var}')
        
        # Subplot 2: Outlier detection (Boxplot)
        plt.subplot(2, 2, 2)
        sns.boxplot(y=df_var)
        plt.title(f'Boxplot of {var}')
        
        # Subplot 3: Normality check (Q-Q plot)
        plt.subplot(2, 2, 3)
        stats.probplot(df_var, dist="norm", plot=plt)
        plt.title(f'Normality Check for {var}')
        
        # Subplot 4: Target relationship (Scatter plot + Pearson correlation)
        plt.subplot(2, 2, 4)
        plt.scatter(df_var, target_filtered, alpha=0.5)
        plt.title(f'Correlation with {target}')
        
        corr = df_var.corr(target_filtered)
        plt.annotate(f'Pearson r: {corr:.2f}', xy=(0.7, 0.9), xycoords='axes fraction',
                     fontsize=12, bbox=dict(boxstyle="round", fc="white"))
        
        plt.suptitle(f'Variable Analysis: {var}', fontsize=16)
        plt.figtext(0.5, 0.92, stats_text, ha='center', va='top', fontsize=12, bbox=dict(facecolor='white', alpha=0.5))
        plt.tight_layout(rect=[0, 0, 1, 0.85])
        plt.show()

# Analysis for discrete variables: value distribution and target impact
def plot_discrete_analysis(df, variables, target):
    for var in variables:
        plt.figure(figsize=(10, 6))
        plt.suptitle(f'Discrete Variable Analysis: {var}', fontsize=16)
        
        # Frequency distribution
        plt.subplot(2, 2, 1)
        order = df[var].value_counts().index
        sns.countplot(x=var, data=df, order=order)
        plt.title(f'Distribution of {var}')
        plt.xticks(rotation=45)
        
        # Target distribution per category
        plt.subplot(2, 2, 2)       
        sns.boxplot(x=var, y=target, data=df, order=order)
        plt.title(f'{target} Distribution by {var}')
        plt.xticks(rotation=45)

        # Print statistics
        total = len(df)
        value_counts = df[var].value_counts(normalize=True).mul(100).round(2)
        missing_pct = (df[var].isna().sum() / total) * 100
        
        print(f"\n{'='*50}\nAnalysis for variable: {var}\n{'-'*50}")
        print("1. Value distribution (% of total):\n", value_counts.to_string())
        print(f"\n2. Missing values: {missing_pct:.2f}%\n{'='*50}")

        plt.tight_layout()
        plt.show()

# Combine features by summation and drop original columns
def sum_and_drop_columns(df, columns_to_sum, new_col_name, drop_original=True):
    missing_cols = [col for col in columns_to_sum if col not in df.columns]
    if missing_cols:
        raise ValueError(f"Columns not found in DataFrame: {missing_cols}")
    
    df[new_col_name] = df[columns_to_sum].sum(axis=1)
    if drop_original:
        df = df.drop(columns=columns_to_sum)
    return df

# Find categorical columns containing the 'Gd' (Good) rating
def find_columns_with_Gd(df, case_sensitive=True):
    categorical_cols = df.select_dtypes(include=['object', 'category']).columns
    Gd_columns = []
    
    for col in categorical_cols:
        unique_vals = df[col].dropna().astype(str).unique()
        has_Gd = 'Gd' in unique_vals if case_sensitive else any(val.upper() == 'GD' for val in unique_vals)
        if has_Gd:
            Gd_columns.append(col)
    return Gd_columns

# Identify imbalanced features where one category dominates >= threshold
def find_dominant_categories(df, threshold=0.9, include_nan=False):
    dominant_vars = []
    dominant_var_names = []
    
    for col in df.columns:
        counts = df[col].value_counts(dropna=not include_nan, normalize=True)
        if not counts.empty and counts.max() >= threshold:
            dominant_var_names.append(col)
            dominant_vars.append({
                'variable': col,
                'dominant_pct': round(counts.max() * 100, 2),
                'dominant_category': counts.idxmax()
            })
    return pd.DataFrame(dominant_vars), dominant_var_names

# Compute Cramer's V to measure correlation between nominal features
def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    return np.sqrt(phi2 / min((k-1), (r-1)))

# Split dataset into train and test sets (70/30)
def split():
    return train_test_split(
        data.drop('SalePrice', axis=1), data['SalePrice'],
        test_size=0.3, random_state=0
    )

# Basic imputation: 0 for numeric, 'none' for categorical variables
def fill_missing_values(df):
    numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
    categorical_cols = df.select_dtypes(include=['object', 'category']).columns
    
    df[numeric_cols] = df[numeric_cols].fillna(0)           
    df[categorical_cols] = df[categorical_cols].fillna("none") 
    return df

# Train and evaluate Linear Regression with standardized features
def run_linear_regression(X_train, X_test, y_train, y_test):
    scaler = StandardScaler(with_mean=False)
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    lr = LinearRegression()
    lr.fit(X_train_scaled, y_train)

    for name, X, y in [("Train set", X_train_scaled, y_train), ("Test set", X_test_scaled, y_test)]:
        print(f"\n{name}")
        pred = lr.predict(X)
        print(f"Linear Regression R2: {r2_score(y, pred):.4f}")
        print(f"Linear Regression MAE: {mean_absolute_error(y, pred):.4f}")
        print(f"Linear Regression RMSE: {np.sqrt(mean_squared_error(y, pred)):.4f}")

# Train and evaluate RandomForestRegressor
def run_randomForests(X_train, X_test, y_train, y_test):
    rf = RandomForestRegressor(n_estimators=100, random_state=0, max_depth=3)
    rf.fit(X_train, y_train)

    for name, X, y in [("Train set", X_train, y_train), ("Test set", X_test, y_test)]:
        print(f"\n{name}")
        pred = rf.predict(X)
        print(f"Random Forest R2: {r2_score(y, pred):.4f}")
        print(f"Random Forest MAE: {mean_absolute_error(y, pred):.4f}")
        print(f"Random Forest RMSE: {np.sqrt(mean_squared_error(y, pred)):.4f}")

# Encode nominal categorical features using One-Hot Encoding
def get_OHE():
    encoder = OneHotEncoder(drop_last=True)
    return encoder.fit_transform(X_train), encoder.transform(X_test)  

# Encode ordinal categorical features using Ordinal Encoding
def get_ordinal(): 
    encoder = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
    return encoder.fit_transform(X_train), encoder.transform(X_test)

# Apply Yeo-Johnson transformation shifted by +1 to normalize distribution
def safe_yeojohnson(x):
    try:
        return yeojohnson(x + 1)[0]
    except:
        print("Yeo-Johnson failed for column, returning original values")
        return x

# Apply Box-Cox transformation shifted by +1 with negative value check
def safe_boxcox(x):
    try:
        x_shifted = x + 1 - np.min(x) if np.min(x) <= 0 else x + 1
        return boxcox(x_shifted)[0]
    except:
        print("Box-Cox failed for column, returning original values")
        return x

# Check distribution normality: histogram and Q-Q plot
def diagnostic_plots(df, variable):
    plt.figure(figsize=(15,6))
    plt.subplot(1,2,1)
    df[variable].hist(bins=30)
    plt.title(variable)

    plt.subplot(1,2,2)
    stats.probplot(df[variable], dist='norm', plot=plt)
    plt.title(variable)
    plt.show()

# Save evaluation metrics (R2, MAE, RMSE) to results DataFrame
def metrics_gridsearch(model_name, y_test, y_train, y_pred_test, y_pred_train, results_df):
    results_df[model_name + '_test'] = [round(r2_score(y_test, y_pred_test), 4), 
                                        round(mean_absolute_error(y_test, y_pred_test), 4), 
                                        round(np.sqrt(mean_squared_error(y_test, y_pred_test)), 4)]
    results_df[model_name + '_train'] = [round(r2_score(y_train, y_pred_train), 4), 
                                         round(mean_absolute_error(y_train, y_pred_train), 4), 
                                         round(np.sqrt(mean_squared_error(y_train, y_pred_train)), 4)]
    return results_df

# Plot train vs test overfit for relative metrics (R2) with % delta
def plot_overfit_with_exact_labels(results_df, metric_name):
    df = results_df.T
    df_metric = df[[metric_name]].copy()
    df_metric['model'] = df_metric.index
    df_metric['type'] = df_metric['model'].apply(lambda x: 'train' if 'train' in x else 'test')
    df_metric['model_clean'] = df_metric['model'].apply(lambda x: x.replace('_train', '').replace('_test', ''))

    pivot = df_metric.pivot(index='model_clean', columns='type', values=metric_name).fillna(0)
    pivot['overfit_%'] = ((pivot['train'] - pivot['test']) / pivot['train']) * 100

    fig, ax = plt.subplots(figsize=(15, 6))
    ax.barh(pivot.index, pivot['train'], color='orange', label='Train')
    ax.barh(pivot.index, pivot['test'], color='steelblue', label='Test')
    
    for i, model in enumerate(pivot.index):
        train_val, test_val, diff = pivot.loc[model, 'train'], pivot.loc[model, 'test'], pivot.loc[model, 'overfit_%']
        ax.text(train_val - 0.01, i - 0.15, f'{train_val*100:.1f}%', color='purple', ha='right', fontsize=12)
        ax.text(test_val - 0.01, i + 0.05, f'{test_val*100:.1f}%', color='blue', ha='right', fontsize=12)
        ax.text(max(train_val, test_val) + 0.02, i, f'{diff:.1f}%', color='darkred', ha='left', va='bottom', fontsize=12)

    ax.set_xlabel(metric_name)
    ax.set_title(f'{metric_name}: Train vs Test with % difference')
    ax.set_xlim(0, 1.1)
    ax.legend()
    ax.grid(axis='x', linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()

# Plot train vs test overfit for regression errors (MAE, RMSE) with % delta
def plot_overfit_with_exact_labels_Reg(results_df, metric_name, top_n=20):
    df = results_df.T
    df_metric = df[[metric_name]].copy()
    df_metric['model'] = df_metric.index
    df_metric['type'] = df_metric['model'].apply(lambda x: 'train' if 'train' in x else 'test')
    df_metric['model_clean'] = df_metric['model'].apply(lambda x: x.replace('_train', '').replace('_test', ''))

    pivot = df_metric.pivot(index='model_clean', columns='type', values=metric_name).fillna(0)
    pivot['overfit_%'] = ((pivot['train'] - pivot['test']) / pivot['train']) * 100
    pivot = pivot.sort_values('train', ascending=True).head(top_n)

    fig, ax = plt.subplots(figsize=(15, top_n * 0.5))
    ax.barh(pivot.index, pivot['test'], color='steelblue', label='Test', zorder=1)
    ax.barh(pivot.index, pivot['train'], color='orange', label='Train', zorder=2)

    for i, model in enumerate(pivot.index):
        train_val, test_val, diff = pivot.loc[model, 'train'], pivot.loc[model, 'test'], pivot.loc[model, 'overfit_%']
        ax.text(train_val + 100, i - 0.15, f'{train_val:.0f}', color='purple', ha='left', fontsize=10)
        ax.text(test_val + 100, i + 0.05, f'{test_val:.0f}', color='blue', ha='left', fontsize=10)
        ax.text(max(train_val, test_val) + 500, i, f'{diff:.1f}%', color='darkred', ha='left', va='bottom', fontsize=10)

    ax.set_xlabel(metric_name)
    ax.set_title(f'{metric_name}: Train vs Test with % difference')
    ax.legend()
    ax.grid(axis='x', linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()
```

## step 2 import Dataset

Ames Housing Dataset Preprocessing
 
This script implements the data cleaning and preprocessing guidelines outlined by Dean De Cock in his paper: 
"Ames, Iowa: Alternative to the Boston Housing Data as an End of Semester Regression Project"  
 (Journal of Statistics Education, 2011).  
https://jse.amstat.org/v19n3/decock.pdf

Key implementations include:
- Removing severe outliers (properties with living area > 4000 sq ft in train data).
- Handling 'NA' structural categories (transforming missing values into 'None' or 0).
- Converting numeric identifiers that represent categorical concepts into strings.
- Dropping zero-variance / uninformative features.
- 'PoolQC': A missing value here is NOT a data loss. According to documentation,'NA' explicitly means "No Pool" (matches PoolArea = 0).  
Replacing with 'None' preserves this as a legitimate categorical feature value.  
- 'GarageYrBlt': Houses with no garage have an empty year field.  
Filling this with 0 would create a massive 2000-year mathematical gap/skewness, severely distorting distance-based and linear models.  
Filling with 'YearBuilt' logically aligns the garage's age profile with the house.  
- 'LotFrontage': Missing values here represent actual data loss or technical omissions (or cul-de-sac properties). Setting it to 0 is physically impossible.  
Since houses in the same neighborhood share similar zoning laws and plot layouts, imputing via the Neighborhood's median provides a highly accurate reconstruction.

* **Per Dean De Cock's Documentation:** The dataset's core task is predicting fair market value. Transactions labeled as `Abnorml` (foreclosures) or `Family` (sales to relatives) occur at prices significantly below true market rates.
* **Filtering Rationale:** Including non-market transactions when training linear models causes weight bias and underestimates high-quality property parameters. Thus, a strict filter for `SaleCondition == 'Normal'` is integrated into the pipeline.

```python
# Data read and view
data = pd.read_csv("houseprice.csv")
data = data.drop(columns='Id')
data.shape
```

```python
na_to_none_cols = [
    'PoolQC', 'MiscFeature', 'Alley', 'Fence', 'FireplaceQu', 
    'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond',
    'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2',
    'MasVnrType'
]

na_to_zero_cols = [
    'GarageArea', 'GarageCars', 'BsmtFinSF1', 'BsmtFinSF2', 
    'BsmtUnfSF', 'TotalBsmtSF', 'BsmtFullBath', 'BsmtHalfBath', 'MasVnrArea'
]

cols_to_stringify = ['MSSubClass', 'MoSold']

if 'GrLivArea' in data.columns:
    data = data[data['GrLivArea'] < 4000].reset_index(drop=True)

for col in na_to_none_cols:
    if col in data.columns:
        data[col] = data[col].fillna('None')

for col in na_to_zero_cols:
    if col in data.columns:
        data[col] = data[col].fillna(0)

if 'GarageYrBlt' in data.columns and 'YearBuilt' in data.columns:
    data['GarageYrBlt'] = data['GarageYrBlt'].fillna(data['YearBuilt'])

if 'LotFrontage' in data.columns and 'Neighborhood' in data.columns:
    data['LotFrontage'] = data.groupby('Neighborhood')['LotFrontage'].transform(lambda x: x.fillna(x.median()))

for col in cols_to_stringify:
    if col in data.columns:
        data[col] = data[col].astype(str)

if 'Utilities' in data.columns:
    data = data.drop(columns=['Utilities'])

data = data[data['SaleCondition'] == 'Normal']
data = data.drop('SaleCondition', axis=1)
data.shape
```
